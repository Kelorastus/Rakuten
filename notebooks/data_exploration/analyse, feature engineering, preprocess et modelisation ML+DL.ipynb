```python
import pandas as pd
import html
import re
import spacy  # Pour la lemmatisation
from nltk.tokenize import word_tokenize
from nltk.tokenize.regexp import RegexpTokenizer
from nltk.corpus import stopwords
import nltk

# Téléchargements NLTK
nltk.download('punkt')
nltk.download('stopwords')

# Charger le modèle spaCy pour le français
nlp = spacy.load("fr_core_news_sm")

# Charger les données
X_fr = pd.read_csv(r"C:\Users\user\Documents\Rakuten\Rakuten\Dataset\X_fr.csv")

# 1. Décoder les entités HTML
X_fr["desc. comp"] = X_fr["desc. comp"].apply(lambda x: html.unescape(str(x)))

# 2. Nettoyage préliminaire (minuscules + contractions)
def nettoyer_texte(texte):
    texte = texte.lower()
    texte = re.sub(r"c'est", "ce_est", texte)
    texte = re.sub(r"l'", "le_", texte)
    texte = re.sub(r"d'", "de_", texte)
    texte = re.sub(r"s'", " ", texte)  # Supprime "s'" (ex: "s'arranger" → " arranger")
    texte = re.sub(r"qu'", "que_", texte)
    texte = re.sub(r"j'", "je_", texte)
    texte = re.sub(r"m'", "me_", texte)
    texte = re.sub(r"t'", "te_", texte)
    return texte

X_fr["desc. comp"] = X_fr["desc. comp"].apply(nettoyer_texte)

# 3. Lemmatisation avec spaCy (normalise les verbes et adjectifs)
def lemmatiser_texte(texte):
    doc = nlp(texte)
    return " ".join([token.lemma_ for token in doc])

X_fr["desc. comp"] = X_fr["desc. comp"].apply(lemmatiser_texte)

# 4. Tokenisation avec word_tokenize
mots = [word_tokenize(desc, language="french") for desc in X_fr["desc. comp"]]

# 5. Suppression des stopwords (liste étendue)
stop_words = set(stopwords.words('french'))
stop_words.update(["c", "j", "d", "l", "m", "t", "qu", "n", "ce", "le_", "de_", "en", "que_", "je_", "me_", "te_"])
mots_sans_stopwords = [
    [word for word in desc if word not in stop_words]
    for desc in mots
]

# 6. Filtrage avec RegexpTokenizer (mots ≥4 lettres)
tokenizer = RegexpTokenizer(r"\b[a-zA-Zàâäéèêëîïôöùûüçœæ'-]{4,}\b", flags=re.IGNORECASE)
tokens_list = [
    tokenizer.tokenize(' '.join(desc))
    for desc in mots_sans_stopwords
]

# Afficher les résultats
print("Exemples de tokens nettoyés et lemmatisés :")
for i, tokens in enumerate(tokens_list[:5], 1):
    print(f"Description {i} : {tokens}")
```

```python
# =============================================
# VECTORISATION TF-IDF OPTIMISÉE
# (À placer APRES ton code de prétraitement existant)
# =============================================

from sklearn.feature_extraction.text import TfidfVectorizer
import joblib
import numpy as np

# 1. Reconstruire les textes à partir des tokens
X_fr["texte_final"] = [" ".join(tokens) for tokens in tokens_list]

# 2. Configuration optimisée du vectoriseur
vectorizer = TfidfVectorizer(
    input='content',            # On travaille sur du texte brut
    encoding='utf-8',           # Pour gérer les caractères spéciaux
    decode_error='replace',     # Remplace les caractères problématiques
    strip_accents='unicode',    # Normalise les accents (é → e)
    lowercase=True,             # Tout en minuscules (déjà fait mais double vérification)
    preprocessor=None,          # Désactive le préprocesseur intégré (on a déjà tout nettoyé)
    tokenizer=None,             # Désactive le tokeniseur intégré (on utilise nos tokens)
    analyzer='word',            # Analyse par mot
    ngram_range=(1, 1),         # Uniquement des unigrammes
    min_df=5,                   # Ignore les mots présents dans <5 descriptions
    max_df=0.95,                # Ignore les mots trop fréquents (>95%)
    max_features=15000,         # Limite le nombre de features
    norm='l2',                  # Normalisation L2 (standard pour les modèles linéaires)
    use_idf=True,               # Active IDF
    smooth_idf=True,            # Lissage IDF
    sublinear_tf=True,          # Applique log(1 + tf) au lieu de tf
    binary=False                # Conserve les comptes TF-IDF (pas binaire)
)

# 3. Vectorisation
print("Début de la vectorisation...")
X_tfidf = vectorizer.fit_transform(X_fr["texte_final"])
print("Vectorisation terminée !")

# 4. Statistiques
print("\n=== STATISTIQUES ===")
print(f"Nombre de descriptions: {X_tfidf.shape[0]}")
print(f"Nombre de features: {X_tfidf.shape[1]}")
print(f"Taille mémoire: {X_tfidf.data.nbytes / (1024**2):.2f} Mo")
print(f"Densité: {X_tfidf.nnz / (X_tfidf.shape[0] * X_tfidf.shape[1]):.2%}")

# 5. Vérification avec un exemple
exemple_texte = "cahier grand format page blanche"
exemple_tokens = tokenizer.tokenize(exemple_texte)  # Utilise TON tokenizer existant
exemple_vectorise = vectorizer.transform([" ".join(exemple_tokens)])

print("\n=== EXEMPLE ===")
print(f"Texte: '{exemple_texte}'")
print(f"Tokens: {exemple_tokens}")
print(f"Somme des valeurs TF-IDF: {exemple_vectorise.sum():.2f}")
print(f"Nombre de features non nulles: {exemple_vectorise.nnz}")

# 6. Top mots pour une description aléatoire
idx_exemple = np.random.randint(0, len(X_fr))
top_features = np.argsort(X_tfidf[idx_exemple].toarray().ravel())[-10:][::-1]
top_mots = [vectorizer.get_feature_names_out()[i] for i in top_features]
top_scores = [X_tfidf[idx_exemple, i] for i in top_features]

print(f"\n=== TOP 10 MOTS (Description {idx_exemple}) ===")
for mot, score in zip(top_mots, top_scores):
    print(f"{mot}: {score:.3f}")
```

```python
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import make_pipeline as make_imb_pipeline
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import SGDClassifier
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.metrics import accuracy_score, f1_score, classification_report
from scipy.sparse import issparse
import numpy as np
from sklearn.base import BaseEstimator, TransformerMixin

# 1. Stratégie SMOTE optimisée (seulement over-sampling pour minorités <800, cible=800)
sampling_strategy = {
    cls: 800
    for cls, n_samples in y_train.value_counts().to_dict().items()
    if n_samples < 800
}

# 2. Classe personnalisée pour sélection de features adaptée aux matrices creuses
class SafeSelectKBest(BaseEstimator, TransformerMixin):
    def __init__(self, k=5000):
        self.k = k
        self.support_ = None

    def fit(self, X, y):
        if issparse(X):
            # Densification partielle pour calculer les scores
            subset_size = min(1000, X.shape[0])
            X_subset = X[:subset_size].toarray()
            y_subset = y[:subset_size]
            scores, _ = f_classif(X_subset, y_subset)
            top_k_indices = np.argsort(scores)[-self.k:]
            self.support_ = top_k_indices
        else:
            selector = SelectKBest(f_classif, k=self.k)
            selector.fit(X, y)
            self.support_ = selector.get_support(indices=True)
        return self

    def transform(self, X):
        if self.support_ is None:
            raise ValueError("Transformer not fitted yet.")
        return X[:, self.support_]

# 3. Pipeline final optimisé (sélection et réduction AVANT SMOTE pour densifier)
optimized_pipeline = make_imb_pipeline(
    # Étape 1: Sélection de features SÛRE
    SafeSelectKBest(k=5000),
    
    # Étape 2: Réduction de dimension (densifie les données)
    TruncatedSVD(
        n_components=1500,
        random_state=42,
        n_iter=20
    ),
    
    # Étape 3: Scaling adapté (sur données denses)
    StandardScaler(with_mean=False),
    
    # Étape 4: SMOTE (maintenant sur données denses et réduites)
    SMOTE(
        sampling_strategy=sampling_strategy,
        random_state=42,
        k_neighbors=3  # Réduit pour éviter le bruit
    ),
    
    # Étape 5: Classifieur optimisé
    SGDClassifier(
        loss='hinge',
        penalty='elasticnet',
        l1_ratio=0.1,
        alpha=0.0001,
        class_weight='balanced',  # Alternative à class_weights
        max_iter=2000,
        tol=1e-4,
        learning_rate='adaptive',
        eta0=0.1,
        random_state=42,
        n_jobs=-1,  # n_jobs est valide ici
        early_stopping=True,
        validation_fraction=0.1
    )
)

# 4. Entraînement avec gestion mémoire
print("🚀 Début de l'entraînement...")
optimized_pipeline.fit(X_train, y_train)
print("✅ Entraînement terminé!")

# 5. Évaluation complète
y_pred = optimized_pipeline.predict(X_test)
print("\n=== MÉTRIQUES CLÉS ===")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print(f"F1 Macro: {f1_score(y_test, y_pred, average='macro'):.4f}")
print(f"F1 Weighted: {f1_score(y_test, y_pred, average='weighted'):.4f}")
print("\n=== RAPPORT COMPLET ===")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
import pandas as pd
import html
import re
import spacy  # Pour la lemmatisation
from nltk.tokenize import word_tokenize
from nltk.tokenize.regexp import RegexpTokenizer
from nltk.corpus import stopwords
import nltk

# Téléchargements NLTK
nltk.download('punkt')
nltk.download('stopwords')

# Charger le modèle spaCy pour l'anglais
nlp = spacy.load("en_core_web_sm")

# Charger les données
X_en = pd.read_csv(r"C:\Users\user\Documents\Rakuten\Rakuten\Dataset\X_en.csv")

# 1. Décoder les entités HTML
X_en["desc. comp"] = X_en["desc. comp"].apply(lambda x: html.unescape(str(x)))

# 2. Nettoyage préliminaire (minuscules + expansions de contractions courantes)
def nettoyer_texte(texte):
    texte = texte.lower()
    # Expansions de contractions courantes en anglais pour aider la lemmatisation
    texte = re.sub(r"it's", "it is", texte)
    texte = re.sub(r"i'm", "i am", texte)
    texte = re.sub(r"don't", "do not", texte)
    texte = re.sub(r"doesn't", "does not", texte)
    texte = re.sub(r"can't", "can not", texte)
    texte = re.sub(r"won't", "will not", texte)
    texte = re.sub(r"you're", "you are", texte)
    texte = re.sub(r"they're", "they are", texte)
    texte = re.sub(r"we're", "we are", texte)
    texte = re.sub(r"he's", "he is", texte)
    texte = re.sub(r"she's", "she is", texte)
    texte = re.sub(r"there's", "there is", texte)
    texte = re.sub(r"that's", "that is", texte)
    # Supprimer les apostrophes restantes ou artefacts (ex: "s'" pour possessifs, mais on les gère simplement)
    texte = re.sub(r"'s", " ", texte)  # Ex: "John's" → "John "
    texte = re.sub(r"'re", " are", texte)
    texte = re.sub(r"'ll", " will", texte)
    texte = re.sub(r"'ve", " have", texte)
    texte = re.sub(r"'d", " would", texte)  # Ou "had", mais "would" est courant
    return texte

X_en["desc. comp"] = X_en["desc. comp"].apply(nettoyer_texte)

# 3. Lemmatisation avec spaCy (normalise les verbes, adjectifs, etc.)
def lemmatiser_texte(texte):
    doc = nlp(texte)
    return " ".join([token.lemma_ for token in doc])

X_en["desc. comp"] = X_en["desc. comp"].apply(lemmatiser_texte)

# 4. Tokenisation avec word_tokenize
mots = [word_tokenize(desc, language="english") for desc in X_en["desc. comp"]]

# 5. Suppression des stopwords (liste étendue pour l'anglais)
stop_words = set(stopwords.words('english'))
# Ajouts personnalisés : mots courts courants ou artefacts potentiels des contractions
stop_words.update(["i", "you", "he", "she", "it", "we", "they", "this", "that", "the", "a", "an", "in", "on", "at", "to", "of", "for", "with", "by"])

mots_sans_stopwords = [
    [word for word in desc if word not in stop_words]
    for desc in mots
]

# 6. Filtrage avec RegexpTokenizer (mots ≥4 lettres, adapté à l'anglais)
tokenizer = RegexpTokenizer(r"\b[a-zA-Z'-]{4,}\b", flags=re.IGNORECASE)
tokens_list = [
    tokenizer.tokenize(' '.join(desc))
    for desc in mots_sans_stopwords
]

# Afficher les résultats
print("Exemples de tokens nettoyés et lemmatisés :")
for i, tokens in enumerate(tokens_list[:5], 1):
    print(f"Description {i} : {tokens}")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score, f1_score
from joblib import dump
import warnings
warnings.filterwarnings("ignore")

# =============================================
# 1. CHARGEMENT DES DONNÉES PRÉTRAITÉES
# =============================================
# On utilise directement la matrice TF-IDF calculée pour les données en anglais
# et les labels correspondants
y_en = pd.read_csv(r"C:\Users\user\Documents\Rakuten\Rakuten\Dataset\y_en.csv")["prdtypecode"]

# Afficher la distribution des classes
print("Distribution des classes:\n", y_en.value_counts().sort_index())

# =============================================
# 2. UTILISATION DE LA MATRICE TF-IDF EXISTANTE
# =============================================
# La matrice X_tfidf a déjà été calculée pour les données en anglais
# On la récupère directement depuis la variable existante
print(f"Taille de la matrice TF-IDF: {X_tfidf.shape}")

# Séparation train/test (stratifiée)
X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf, y_en,
    test_size=0.2,
    random_state=42,
    stratify=y_en
)
print(f"Taille X_train: {X_train.shape}, y_train: {len(y_train)}")

# =============================================
# 3. CALCUL DES POIDS DE CLASSE OPTIMISÉS
# =============================================
# Calcul automatique basé sur la distribution réelle
class_counts = y_en.value_counts()
majority_count = class_counts.max()
class_weights = {
    c: min(majority_count / count, 5.0)  # Plafond à 5.0
    for c, count in class_counts.items()
}
print("\nPoids des classes (exemples):")
for c in [10, 1180, 2583, 1280, 1302]:
    print(f"Classe {c}: poids = {class_weights.get(c, 1.0):.2f}")

# =============================================
# 4. DÉFINITION DES MODÈLES OPTIMISÉS (AJOUT DE SGDClassifier AVEC HYPERPARAMÈTRES ÉTENDUS)
# =============================================
models = {
    "LinearSVC": {
        "model": LinearSVC(
            random_state=42,
            dual=False,  # Important pour n_samples > n_features
            max_iter=5000  # Augmenté pour convergence
        ),
        "params": {
            "C": [0.1, 1, 10],
            "class_weight": [class_weights]
        }
    },
    "LogisticRegression": {
        "model": LogisticRegression(
            max_iter=1000,
            random_state=42,
            solver="saga",  # Meilleur pour les grands datasets
            n_jobs=-1  # Utilise tous les cœurs
        ),
        "params": {
            "C": [0.1, 1, 10],
            "class_weight": [class_weights]
        }
    },
    "SGDClassifier": {
        "model": SGDClassifier(
            random_state=42,
            n_jobs=-1,  # Utilise tous les cœurs
            early_stopping=True,
            validation_fraction=0.1
        ),
        "params": {
            "loss": ['hinge', 'log_loss', 'modified_huber'],
            "penalty": ['l2', 'l1', 'elasticnet'],
            "alpha": [1e-5, 1e-4, 1e-3],
            "l1_ratio": [0.1, 0.15, 0.2],
            "learning_rate": ['optimal', 'adaptive'],
            "eta0": [0.01, 0.1],
            "max_iter": [1000, 2000],
            "tol": [1e-4],
            "class_weight": [class_weights, 'balanced']
        }
    }
}

# =============================================
# 5. OPTIMISATION AVEC GRIDSEARCHCV
# =============================================
results = {}
for name, config in models.items():
    print(f"\n=== Optimisation {name} ===")
    grid = GridSearchCV(
        estimator=config["model"],
        param_grid=config["params"],
        cv=StratifiedKFold(n_splits=3),  # 3 folds pour un bon compromis
        scoring="f1_macro",  # Métrique optimisée pour le déséquilibre
        n_jobs=-1,  # Utilise tous les cœurs
        verbose=2  # Plus de détails sur la progression
    )
    grid.fit(X_train, y_train)
    results[name] = {
        "model": grid.best_estimator_,
        "best_params": grid.best_params_,
        "best_score": grid.best_score_
    }
    print(f"Meilleurs paramètres: {grid.best_params_}")
    print(f"Score F1 moyen (CV): {grid.best_score_:.3f}")

# =============================================
# 6. ÉVALUATION COMPLÈTE SUR LE TEST SET
# =============================================
print("\n=== PERFORMANCES SUR LE TEST SET ===")
final_results = {}
for name, data in results.items():
    y_pred = data["model"].predict(X_test)
    final_results[name] = {
        "accuracy": accuracy_score(y_test, y_pred),
        "f1_macro": f1_score(y_test, y_pred, average="macro"),
        "f1_weighted": f1_score(y_test, y_pred, average="weighted"),
        "classification_report": classification_report(y_test, y_pred, zero_division=0)
    }
    print(f"\n--- {name} ---")
    print(f"Accuracy: {final_results[name]['accuracy']:.3f}")
    print(f"F1 Macro: {final_results[name]['f1_macro']:.3f}")
    print(f"F1 Weighted: {final_results[name]['f1_weighted']:.3f}")
    print("Classification Report:")
    print(final_results[name]["classification_report"])

# =============================================
# 7. ANALYSE DES RÉSULTATS
# =============================================
results_df = pd.DataFrame({
    "Modèle": list(final_results.keys()),
    "Accuracy": [r["accuracy"] for r in final_results.values()],
    "F1 Macro": [r["f1_macro"] for r in final_results.values()],
    "F1 Weighted": [r["f1_weighted"] for r in final_results.values()],
    "Meilleurs Paramètres": [str(results[n]["best_params"]) for n in final_results.keys()]
}).sort_values(by="F1 Macro", ascending=False)

print("\n=== RÉCAPITULATIF ===")
print(results_df)

# =============================================
# 8. SAUVEGARDE DU MEILLEUR MODÈLE
# =============================================
best_model_name = results_df.iloc[0]["Modèle"]
best_model = results[best_model_name]["model"]

# Sauvegarde complète (modèle + vectorizer + poids)
dump({
    "model": best_model,
    "vectorizer": vectorizer,  # Le vectorizer utilisé pour les données en anglais
    "class_weights": class_weights,
    "best_params": results[best_model_name]["best_params"]
}, "meilleur_modele_optimise_en.pkl")

print(f"\n✅ Meilleur modèle ({best_model_name}) sauvegardé dans 'meilleur_modele_optimise_en.pkl'")
print("Configuration optimale:")
print(f"- C: {results[best_model_name]['best_params'].get('C', 'N/A')}")
print(f"- Class weights: personnalisés (max weight: {max(class_weights.values()):.1f})")
print(f"- Score F1 Macro: {final_results[best_model_name]['f1_macro']:.3f}")
```

```python
import os
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms
from PIL import Image
from sklearn.model_selection import train_test_split
from transformers import AutoTokenizer
from torch.cuda.amp import autocast, GradScaler
from tqdm import tqdm

# ==========================================
# 1. CONFIGURATION ET DISPOSITIF
# ==========================================
# Remplace par ton chemin réel si nécessaire
img_dir = r"C:\Users\user\Downloads\images(1)\images\image_train"
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ==========================================
# 2. DATASET MULTIMODAL
# ==========================================
class RakutenMultimodalDataset(Dataset):
    def __init__(self, dataframe, img_dir, tokenizer, max_len=64):
        self.df = dataframe.reset_index(drop=True)
        self.img_dir = img_dir
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.transform = transforms.Compose([
            transforms.Resize((128, 128)), # Légèrement plus grand pour ResNet
            transforms.ToTensor(),
            transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
        ])

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        
        # Image : Nom du fichier basé sur tes colonnes
        filename = f"image_{int(row['imageid'])}_product_{int(row['productid'])}.jpg"
        img_path = os.path.join(self.img_dir, filename)
        
        try:
            image = Image.open(img_path).convert("RGB")
            image = self.transform(image)
        except Exception as e:
            # En cas d'image corrompue, on renvoie un tenseur vide (ou gère l'erreur)
            image = torch.zeros((3, 128, 128))
        
        # Texte : Utilisation de ta colonne 'desc. comp'
        text = str(row["desc. comp"])
        tokens = self.tokenizer(
            text, 
            truncation=True, 
            max_length=self.max_len, 
            padding='max_length', 
            return_tensors="pt"
        )
        
        return image, tokens['input_ids'].squeeze(0), torch.tensor(row["label_idx"], dtype=torch.long)

# ==========================================
# 3. MODÈLE DE FUSION (IMAGE + TEXTE)
# ==========================================
class MultimodalModel(nn.Module):
    def __init__(self, num_classes, vocab_size, embed_dim=128):
        super().__init__()
        # Branche Image : ResNet50 pré-entraîné
        self.resnet = models.resnet50(weights=models.ResNet50_Weights.IMAGENET1K_V1)
        self.resnet.fc = nn.Identity() # On garde les features (2048)
        
        # Branche Texte : EmbeddingBag (efficace et rapide)
        self.embedding = nn.EmbeddingBag(vocab_size, embed_dim)
        
        # Classifieur Final
        self.classifier = nn.Sequential(
            nn.Linear(2048 + embed_dim, 512),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(512, num_classes)
        )

    def forward(self, image, text_indices):
        img_feats = self.resnet(image)             # [batch, 2048]
        txt_feats = self.embedding(text_indices)    # [batch, 128]
        
        # Concaténation des deux mondes
        fusion = torch.cat([img_feats, txt_feats], dim=1)
        return self.classifier(fusion)

# ==========================================
# 4. SCRIPT PRINCIPAL
# ==========================================
if __name__ == '__main__':
    # Initialisation du Tokenizer
    tokenizer = AutoTokenizer.from_pretrained("bert-base-multilingual-cased")
    num_classes = len(X["label_idx"].unique())

    # Split Train/Val
    train_df, val_df = train_test_split(X, test_size=0.1, stratify=X["label_idx"], random_state=42)

    # DataLoaders
    train_loader = DataLoader(
        RakutenMultimodalDataset(train_df, img_dir, tokenizer), 
        batch_size=64, shuffle=True, num_workers=0
    )

    # Initialisation Modèle, Optimiseur et Scaler
    model = MultimodalModel(num_classes, tokenizer.vocab_size).to(device)
    optimizer = optim.Adam(model.parameters(), lr=1e-4)
    criterion = nn.CrossEntropyLoss()
    scaler = GradScaler()

    print(f"🚀 Décollage sur {device} avec {len(train_df)} exemples...")

    # Boucle d'entraînement
    epochs = 3
    for epoch in range(epochs):
        model.train()
        loop = tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}")
        
        for images, text_indices, labels in loop:
            images, text_indices, labels = images.to(device), text_indices.to(device), labels.to(device)

            optimizer.zero_grad()
            
            with autocast():
                outputs = model(images, text_indices)
                loss = criterion(outputs, labels)

            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()

            loop.set_postfix(loss=f"{loss.item():.4f}")

    # Sauvegarde finale
    torch.save(model.state_dict(), "rakuten_multimodal_final.pt")
    print("✅ Entraînement terminé et modèle sauvegardé !")
```

```python
# Recréation du val_loader si nécessaire
val_df = train_test_split(X, test_size=0.1, stratify=X["label_idx"], random_state=42)[1]

val_loader = DataLoader(
    RakutenMultimodalDataset(val_df, img_dir, tokenizer), 
    batch_size=64, 
    shuffle=False, 
    num_workers=0
)
```

```python
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

model.eval()
y_true = []
y_pred = []

print("Calcul des scores sur le set de validation...")
with torch.no_grad():
    for images, text_indices, labels in tqdm(val_loader):
        images, text_indices = images.to(device), text_indices.to(device)
        
        outputs = model(images, text_indices)
        _, predicted = torch.max(outputs, 1)
        
        y_true.extend(labels.numpy())
        y_pred.extend(predicted.cpu().numpy())

# Conversion en tableaux numpy pour sklearn
y_true = np.array(y_true)
y_pred = np.array(y_pred)
```

```python
# On utilise class2idx pour avoir les vrais noms de catégories si possible
target_names = [str(c) for c in classes] 
print("\n--- Rapport de Classification ---")
print(classification_report(y_true, y_pred, target_names=target_names))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(12, 10))
sns.heatmap(cm, annot=False, fmt='d', cmap='Blues', 
            xticklabels=target_names, yticklabels=target_names)
plt.xlabel('Prédictions')
plt.ylabel('Vrais Labels')
plt.title('Matrice de Confusion')
plt.show()
```
